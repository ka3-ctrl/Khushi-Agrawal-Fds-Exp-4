```python
import pandas as pd
import numpy as np
from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))

# Load only the columns needed for this experiment
columns = [
    "avg_temp",
    "min_temp",
    "max_temp",
    "wind_speed",
    "air_pressure"
]

df = pd.read_excel(filename, usecols=columns)

# Convert selected columns to numeric values
for col in columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("Dataset imported successfully!")
print(df.head())

print("\nMissing values in each column:")
print(df.isnull().sum())
```

```python

def train_regression(X, y):
    X = np.column_stack((np.ones(len(X)), X))
    coefficients = np.linalg.pinv(X.T @ X) @ X.T @ y

    return coefficients
def predict_regression(X, coefficients):
    X = np.column_stack((np.ones(len(X)), X))
    return X @ coefficients
def evaluate_model(actual, predicted):
    mae = np.mean(np.abs(actual - predicted))
    rmse = np.sqrt(np.mean((actual - predicted) ** 2))

    ss_total = np.sum((actual - np.mean(actual)) ** 2)
    ss_error = np.sum((actual - predicted) ** 2)

    r2 = 1 - ss_error / ss_total if ss_total != 0 else np.nan

    print("MAE:", round(mae, 3))
    print("RMSE:", round(rmse, 3))
    print("R-squared:", round(r2, 3))
```

```python

simple_data = df[["avg_temp", "max_temp"]].dropna()

X = simple_data[["max_temp"]].to_numpy()
y = simple_data["avg_temp"].to_numpy()

np.random.seed(42)
indices = np.random.permutation(len(y))
split = int(0.8 * len(y))

train_idx = indices[:split]
test_idx = indices[split:]

simple_coefficients = train_regression(
    X[train_idx], y[train_idx]
)

simple_predictions = predict_regression(
    X[test_idx], simple_coefficients
)

print("SIMPLE LINEAR REGRESSION")
print("Intercept:", round(simple_coefficients[0], 4))
print("Slope:", round(simple_coefficients[1], 4))

print("\nPrediction accuracy:")
evaluate_model(y[test_idx], simple_predictions)
```

```python
features = [
    "min_temp",
    "max_temp",
    "wind_speed",
    "air_pressure"
]

multiple_data = df[["avg_temp"] + features].dropna()

X_multi = multiple_data[features].to_numpy()
y_multi = multiple_data["avg_temp"].to_numpy()

np.random.seed(42)
indices = np.random.permutation(len(y_multi))
split = int(0.8 * len(y_multi))

train_idx = indices[:split]
test_idx = indices[split:]

multiple_coefficients = train_regression(
    X_multi[train_idx], y_multi[train_idx]
)

multiple_predictions = predict_regression(
    X_multi[test_idx], multiple_coefficients
)

print("MULTIPLE LINEAR REGRESSION")
print("Intercept:", round(multiple_coefficients[0], 4))

for name, coefficient in zip(features, multiple_coefficients[1:]):
    print(name, "coefficient:", round(coefficient, 4))

print("\nPrediction accuracy:")
evaluate_model(y_multi[test_idx], multiple_predictions)
```

```python
missing_mask = df["avg_temp"].isna()

simple_mask = missing_mask & df["max_temp"].notna()

df["simple_predicted_avg_temp"] = np.nan

if simple_mask.any():
    X_missing_simple = df.loc[
        simple_mask, ["max_temp"]
    ].to_numpy()

    df.loc[simple_mask, "simple_predicted_avg_temp"] = (
        predict_regression(X_missing_simple, simple_coefficients)
    )
multiple_mask = missing_mask & df[features].notna().all(axis=1)

df["multiple_predicted_avg_temp"] = np.nan

if multiple_mask.any():
    X_missing_multi = df.loc[
        multiple_mask, features
    ].to_numpy()

    df.loc[multiple_mask, "multiple_predicted_avg_temp"] = (
        predict_regression(X_missing_multi, multiple_coefficients)
    )

print("Missing average temperature records:", int(missing_mask.sum()))
print("Simple regression predictions:", int(simple_mask.sum()))
print("Multiple regression predictions:", int(multiple_mask.sum()))

print("\nPredicted values:")
print(df.loc[missing_mask].head(10))

output_file = "weather_regression_predictions.csv"
df.to_csv(output_file, index=False)

files.download(output_file)
```
